## 3. Custom Tools
We built two custom tools for the Agent:
- `search_web`: Searches the internet for fresh data like GDP, news, and facts. The Agent uses it when it does not know the answer.
- `calculate`: Executes Python code to do math calculations accurately, instead of guessing the result.

```python
@tool
def search_web(query: str) -> str:
    """Search the web for fresh information."""
    try:
        from ddgs import DDGS
        with DDGS() as ddgs:
            results = list(ddgs.text(query, max_results=5))
        if not results:
            return "No results found, use your general knowledge."
        return "\n".join([f"{r['title']}: {r['body']} - {r['href']}" for r in results])
    except Exception as e:
        return f"Search failed, use your general knowledge. Error: {e}"
```
